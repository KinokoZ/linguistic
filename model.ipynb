```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/Users/fengqiuyi/Downloads/Homework/MA675/consult10.2/perception_complete.csv")
df
```

```python
df_ceb=df[df['lang_source']=='ceb']
df_spn=df[df['lang_source']=='spn']
df_ceb_click=df_ceb[df['target']=='click']
df_ceb_knock=df_ceb[df['target']=='knock']
df_spn_click=df_spn[df['target']=='click']
df_spn_knock=df_spn[df['target']=='knock']
```

```python
df_ceb_click_avg=df_ceb_click.groupby("vk")['rating'].mean()
df_ceb_knock_avg=df_ceb_knock.groupby('vk')['rating'].mean()
df_spn_click_avg=df_spn_click.groupby("vk")['rating'].mean()
df_spn_knock_avg=df_spn_knock.groupby('vk')['rating'].mean()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.style.use('ggplot')
```

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 10))
axs[0, 0].bar(df_ceb_click_avg.index, df_ceb_click_avg.values,color='skyblue')
axs[0, 0].set_title('Average Rating by VK')
axs[0, 0].set_xlabel('Ceb Clivk')
axs[0, 0].set_ylabel('Average Rating')

axs[0, 1].bar(df_ceb_knock_avg.index, df_ceb_knock_avg.values,color='salmon')
axs[0, 1].set_title('Average Rating by VK')
axs[0, 1].set_xlabel('Ceb Knock')
axs[0, 1].set_ylabel('Average Rating')

axs[1, 0].bar(df_spn_click_avg.index, df_spn_click_avg.values,color='skyblue')
axs[1, 0].set_title('Average Rating by VK')
axs[1, 0].set_xlabel('Spn Clivk')
axs[1, 0].set_ylabel('Average Rating')

axs[1, 1].bar(df_spn_knock_avg.index, df_spn_knock_avg.values,color='salmon')
axs[1, 1].set_title('Average Rating by VK')
axs[1, 1].set_xlabel('Spn Knock')
axs[1, 1].set_ylabel('Average Rating')
```

```python

fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# First subplot
axs[0].plot(df_ceb_click_avg.index, df_ceb_click_avg.values, color='skyblue', label='Click Average')
axs[0].plot(df_ceb_knock_avg.index, df_ceb_knock_avg.values, color='salmon', label='Knock Average')
axs[0].set_title('Ceb')
axs[0].set_ylabel('Average Rating')
axs[0].set_ylim(1, 10)
axs[0].legend(loc='upper left')  # Add legend to the first subplot

# Second subplot
axs[1].plot(df_spn_click_avg.index, df_spn_click_avg.values, color='skyblue', label='Click Average')
axs[1].plot(df_spn_knock_avg.index, df_spn_knock_avg.values, color='salmon', label='Knock Average')
axs[1].set_title('Spn')
axs[1].set_ylabel('Average Rating')
axs[1].set_ylim(1, 10)
axs[1].legend(loc='upper left')  # Add legend to the second subplot

# Show the plot
plt.tight_layout()  # Adjust layout to prevent overlap
plt.show()
```
